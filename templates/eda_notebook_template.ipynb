## Question 3: Distributions of numeric variables

```python
numeric = raw.select_dtypes('number')
n = max(len(numeric.columns), 1)
fig, axes = plt.subplots(1, n, figsize=(4 * n, 3), squeeze=False)
for ax, col in zip(axes[0], numeric.columns):
    ax.hist(numeric[col].dropna(), bins=30)
    ax.set(title=col, xlabel=f'{col} (units: ?)', ylabel='count of rows')
fig.tight_layout()
plt.show()
```
